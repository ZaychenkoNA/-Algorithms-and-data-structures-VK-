**1. Найти корень числа (ближайшее целое)**

Необходимо написать функцию, которая находит целую часть квадратного корня числа
(округление вниз).

Реализуйте решение с использованием бинарного поиска без использования
встроенных функций извлечения корня. Объясните, почему алгоритм работает за O(log
n) и как определяется граница поиска.


```python
def findSqrt(num):
    left = 0
    right = num
    while left <= right:
        middle = (left + right) // 2
        if middle ** 2 > num:
            right = middle - 1
            continue
        if middle ** 2 < num:
            left = middle + 1
            continue
        return middle
    return right
print(findSqrt(2501))
print(findSqrt(3691))


```

Алгоритм основан на бинарном поиске, сложность которого О(log N). Граница поиска от нуля до самого числа.

**2. Очень лёгкая задача (два ксерокса)**

Как быстро можно сделать N копий документа, если есть два ксерокса, каждый
копирует с разной скоростью — x и y минут за копию?

Реализуйте алгоритм нахождения минимального времени, за которое будет сделано N
копий. Используйте бинарный поиск по времени и поясните, как проверяется условие
достаточности времени.


```python
def copyTime(n, x, y):
    left = 0
    right = (n - 1) * max(x, y)
    while left < right - 1:
        middle = (right + left) // 2
        if middle // x + middle // y < n - 1:
            left = middle
        else:
            right = middle
    return right + min(x, y)

copyTime(30, 2, 3)

```

Достаточность времени t + min(x, y) проверяется условием:

 t // x + t // y >= n-1 (первую копию печатаем на быстром принтере)

**3. Накормить животных**

В зоопарке есть животные, каждое из которых требует определённое количество еды
(целое число).
Каждая привезённая порция еды может накормить только одно животное и не
делится.
Необходимо определить, сколько животных можно накормить имеющимся набором
порций.

Реализуйте жадный алгоритм распределения еды (например, сортировка животных и
порций). Объясните, почему выбранная стратегия даёт максимальное количество
накормленных животных.

```python
def max_fed_animals(animals, portions):
    animals.sort()
    portions.sort()
    i = 0
    j = 0
    fed_count = 0
    while i < len(animals) and j < len(portions):
        if portions[j] >= animals[i]:
            fed_count += 1
            i += 1
            j += 1
        else:
            j += 1
    return fed_count
print(max_fed_animals([4, 2, 7], [3, 5, 2, 6]))
```

Под большими и меньшими животными подразумевается животное, которое требует большее и меньшее количество еды соответственно.
1. Почему нужно кормить сначала самых маленьких животных?

Предположим, существует некое оптимальное решение, которое кормит большее
животное, но не кормит меньшее животное. Тогда мы можем отдать порцию для большего животного меньшему.
Количество накормленных животных не изменится.

2. Почему нужно отдавать самую маленькую подходящую порцию?

Пусть, мы покормили какое-то животное. Если его можно покормить порцией меньше, то сделаем это. Количество накормленных животных не изменится, а освободившейся порцией можно будет покормить животное побольше (если такое найдётся)

**4. Найти разницу между двумя строками**

Даны строки a и b.
Строка b получена из строки a путём перемешивания символов и добавления одной
новой буквы.
Необходимо вернуть эту добавленную букву.

Реализуйте решение без сортировки строк (желательно за O(n)). Объясните, как
используется подсчёт символов или XOR для нахождения лишнего элемента.

```python
def find_added_char(a, b):
    counts = {}
    for ch in a:
        counts[ch] = counts.get(ch, 0) + 1

    for ch in b:
        counts[ch] = counts.get(ch, 0) - 1

    for ch, diff in counts.items():
        if diff != 0:
            return ch
    return ''
print(find_added_char('asdf', 'dsfha'))
print(find_added_char('asdf', 'dsfa'))
```

Каждое включение символа в строку b увеличивает значение у соответствующего ключа, а в строку a - уменьшает. Отсюда получим добавленный символ.

**5. Сумма двух элементов массива**

Дан неотсортированный массив целых чисел и число target.
Необходимо найти два элемента массива, сумма которых равна target.
Каждый элемент можно использовать только один раз.
Если подходящей пары нет — вернуть пустой массив.

Реализуйте эффективное решение за O(n), используя хеш-таблицу (словарь).
Поясните, как хранение уже просмотренных элементов позволяет избежать двойного
цикла.


```python
def twoSum(arr, target):
    seen = {}
    for i, num in enumerate(arr):
        compl = target - num
        if compl in seen:
            return [seen[compl], i]
        seen[num] = i
    return []
print(twoSum([3, 8, 9, 11, 16, 18, 19, 21], 25))
print(twoSum([3, 8, 9, 11, 16, 18, 19, 21], 100))
```

Вместо того, чтобы использовать двойной цикл, мы смотрим пару в хеш-таблице, обращение к которой выполняется за O(1).

**6. Массив анаграмм**

Дан массив строк strs. Сгруппируйте анаграммы вместе. Вы можете вернуть ответ в
любом порядке.

```python
# Способ 1: через сортировки строк
def group_anagrams_sort(strs):
    groups = {}
    for s in strs:
        key = "".join(sorted(s))
        if key not in groups:
            groups[key] = []
        groups[key].append(s)
    return list(groups.values())

print(group_anagrams_sort(["eat","tea","tan","ate","nat","bat"]))

# Сложность O(N * K * logK), где N - число строк, K - максимальная длина строки
```

```python
# Способ 2: без сортировок
def group_anagrams(strs):
    groups = {}
    for s in strs:
        count = [0] * 26
        for ch in s:
            count[ord(ch) - ord('a')] += 1
        key = tuple(count)
        if key not in groups:
            groups[key] = []
        groups[key].append(s)

    return list(groups.values())
print(group_anagrams(["eat","tea","tan","ate","nat","bat"]))

# Сложность O(N * K), однако в этом способе ограничение на символы строки: буквы латинского алфавита
```
