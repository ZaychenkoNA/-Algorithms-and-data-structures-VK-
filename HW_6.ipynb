**1. Наибольшая непрерывная возрастающая последовательность**

Дан массив не отсортированных чисел. Необходимо найти максимально длинную
непрерывную возрастающую последовательность и вернуть её длину.

Проанализируйте решение за O(n) и объясните, как отслеживается текущая и
максимальная длина последовательности.


```python
def findLIS(arr):
    if len(arr) == 0:
        return 0
    if len(arr) == 1:
        return 1
    dp = [1] * len(arr)
    for i in range(1, len(arr)):
        if arr[i - 1] < arr[i]:
            dp[i] = dp[i - 1] + 1
    return max(dp)

print(findLIS([1, 2, 3, 7, 4, 5, 6, 7, 8, 10]))
```

Инициализируем массив dp из единичек, длина которого равна длине массива на входе. Если последовательность продолжает возрастать, то меняем dp[i] на dp[i - 1] + 1. Таким образом запоминаем текущую длину последовательности. Максимальную находим за проход по массиву dp.

**2. Треугольник Паскаля**

Дано число n. Необходимо построить треугольник Паскаля из n строк.
Объясните, как формируется каждая строка на основе предыдущей и какова
сложность алгоритма.

```python
def PascalTriangle(n):
    dp = []
    for i in range(n):
        tmp = []
        for j in range(i + 1):
            tmp.append(1)
        dp.append(tmp)

    for row in range(2, n):
        for col in range(1, row):
            dp[row][col] = dp[row-1][col-1] + dp[row-1][col]
    return dp
print(PascalTriangle(6))
```

Элемент текущей строки с индексом i равен сумме элементов предыдущей строки с индексами i-1 и i (кроме крайних элементов строки - они остаются равными 1).

Сложность по памяти О(n^2) - нужно держать в памяти весь треугольник.

Сложность по времени О((n+1)(n)/2) + O((n-1)(n-2)/2) = O(n^2) - по времени (инициализация треугольника из единиц + перезапись не крайних элементов)

**3. Размен монет**

Дан массив номиналов монет coins и число amount. Требуется найти минимальное
количество монет, составляющих сумму amount, либо вернуть -1, если это
невозможно.

Реализуйте решение с использованием динамического программирования и
обоснуйте корректность переходов.


```python
def coinChange(coins, amount):
    dp = []
    for i in range(amount + 1):
        dp.append(10 ** 50)
    dp[0] = 0
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)
    if (dp[amount] == 10 ** 50):
        return -1
    return dp[amount]
print(coinChange([1, 2, 5], 24))
print(coinChange([2, 5], 3))
```

Заполняем массив dp - количество монет, которое нужно для размена чисел от 0 до amount. Пусть все значения от 0 до i - 1 уже заполнены. Тогда dp[i] = min(dp[i], dp[i - coin] + 1) для каждой монеты coin из массива coins. Так по индукции заполним весь массив и дойдём до dp[amount].
