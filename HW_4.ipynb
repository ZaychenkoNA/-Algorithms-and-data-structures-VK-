Сначала создадим класс узла бинарного дерева.

```python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

**1. Восстановление бинарного дерева из массива**

Дан массив, представляющий бинарное дерево (например, в виде уровнями —
level-order).
Необходимо восстановить структуру бинарного дерева.

Реализуйте функцию, которая строит дерево по переданному массиву. Объясните, как
определяется связь родитель–потомок и какая структура данных используется для
хранения узлов.


```python
def build_tree(arr, i):
    if i >= len(arr):
        return None
    if arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)
    return root
```

Для полного бинарного дерева с индексацией с 0 справедливо свойство:

parent(i) = [i-1/2] (целая часть снизу),

left(i)=2i+1,

right(i)=2i+2.

Структура для хранения узлов в самом начале - через указатели.

**2. Симметричное бинарное дерево**

Дано бинарное дерево.
Необходимо определить, является ли оно симметричным относительно своей
середины.

Реализуйте рекурсивную или итеративную проверку зеркальности левого и правого
поддеревьев. Поясните, какие пары узлов сравниваются и почему алгоритм работает
корректно.


```python
def is_symmetric(root):
    if root is None:
        return True

    def is_mirror(left, right):
        if left is None and right is None:
            return True
        if left is None or right is None:
            return False
        return (left.val == right.val and
                is_mirror(left.left, right.right) and
                is_mirror(left.right, right.left))
    return is_mirror(root.left, root.right)
```

Дерево симметрично <=> левое и правое поддерево являются зеркальным отражением друг друга. Проверяем значения в зеркальных узлах (они должны быть равны) и рекурсивно вызываем проверку для каждого поддерева ниже.

**3. Поиск минимальной глубины бинарного дерева**

Дано бинарное дерево.
Необходимо найти его минимальную глубину (расстояние от корня до ближайшего
листа).

Реализуйте решение с использованием BFS или DFS. Объясните, почему поиск в
ширину (BFS) позволяет корректно найти минимальную глубину.

```python
def minDepth_dfs(root):
    if root is None:
        return 0

    if (root.left is None and root.right is None):
        return 1

    if (root.left is not None and root.right is not None):
        return 1 + min(minDepth_dfs(root.left), minDepth_dfs(root.right))

    if root.left is not None:
        return 1 + minDepth_dfs(root.left)

    if root.right is not None:
        return 1 + minDepth_dfs(root.right)
```

```python
from collections import deque

def minDepth_bfs(root):
    if root is None:
        return 0

    queue = deque([(root, 1)])

    while queue:
        node, depth = queue.popleft()

        if node.left is None and node.right is None:
            return depth

        if node.left:
            queue.append((node.left, depth + 1))
        if node.right:
            queue.append((node.right, depth + 1))
```

BFS обходит дерево послойно (сначала все узлы глубины 1, потом глубины 2, и т.д.). Следовательно, первый встреченный лист гарантированно находится на минимальной глубине.

**5. Сравнение двух бинарных деревьев**

Даны два бинарных дерева.
Необходимо определить, являются ли они идентичными (по структуре и значениям
узлов).

Реализуйте рекурсивную проверку соответствия узлов обоих деревьев. Объясните,
какие условия должны выполняться, чтобы деревья считались одинаковыми.

```python
def isSameTree(a, b):
    if a is None and b is None:
        return True
    if a is None or b is None:
        return False
    if a.val != b.val:
        return False
    return isSameTree(a.left, b.left) and isSameTree(a.right, b.right)
```

Деревья одинаковы, если рекурсивно выполняется:  оба корня пусты, либо оба корня не пусты, их значения равны, и левые/правые поддеревья попарно одинаковы.
