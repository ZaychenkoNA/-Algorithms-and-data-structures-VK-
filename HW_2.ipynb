Сначала реализуем односвязный список для проверки будущих задач. Нам будет достаточно инициализации списка, добавление элемента в конец, и вывода списка. Также для удобства напишем функцию превращения обычного списка в односвязный.

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        new_node = Node(value)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node


def list_to_linked_list(values):
    if not values:
        return None

    head = Node(values[0])
    current = head

    for value in values[1:]:
        current.next = Node(value)
        current = current.next

    return head

def print_linked_list(head):
    current = head
    while current:
        print(current.value, end=" -> ")
        current = current.next
    print("None")
```

**1. Развернуть односвязный список**

Необходимо написать функцию, которая принимает на вход односвязный список и
разворачивает его.


Реализуйте разворот списка in-place, изменяя ссылки между узлами. Объясните, как
меняются указатели на каждом шаге и почему алгоритм работает за O(n).

```python
def reverseLinkedList(head):
    prev = None
    current = head
    while current:
        next = current.next
        current.next = prev
        prev = current
        current = next

    head = prev
    return head

linked_list = list_to_linked_list([1, 2, 3, 4, 5])
print_linked_list(linked_list)
reversed = reverseLinkedList(linked_list)
print_linked_list(reversed)
```

prev - указывает на предыдущий элемент

current - на текущий элемент

next - на следующий (все в изначальном порядке, до разворота)

После того, как мы развернули ссылку, сдвигаем все указатели на один вперед (по изначальному порядку). Алгоритм работает за один проход по списку, т.е. за О(N)

**2. Найти середину списка**

Дан односвязный список.
Необходимо найти его середину за O(n) без дополнительных аллокаций.

Используйте метод двух указателей (медленный и быстрый). Поясните, почему
быстрый указатель позволяет определить середину за один проход.

```python
def middleNode(head):
    slow = head
    fast = head
    while (fast != None) and (fast.next != None):
        slow = slow.next
        fast = fast.next.next
    return slow.value
linked_list = list_to_linked_list([1, 2, 3, 4, 5])
print(middleNode(linked_list))
```

Быстрый указатель движется в 2 раза быстрее. Поэтому, когда он достигнет конца, медленный указатель окажется на середине списка.

**3. Удалить элемент из односвязного списка**

Необходимо написать функцию, которая принимает односвязный список и целое число
val. Нужно удалить узел, значение которого равно val.

Реализуйте удаление узла без создания нового списка. Объясните, как
обрабатываются случаи удаления головы списка и внутренних элементов.


```python
def removeElement(head, val):
    dummy = Node(0)
    dummy.next = head
    prev = dummy
    current = head
    while current:
        if current.value == val:
            prev.next = current.next
        else:
            prev = current
        current = current.next
    return dummy.next
head = list_to_linked_list([4, 1, 2, 1, 3, 1])
head = removeElement(head, 6)
print_linked_list(head)
head = removeElement(head, 1)
print_linked_list(head)
head = removeElement(head, 4)
print_linked_list(head)

```

Благодаря фиктивному узлу dummy, который помещается в начало списка, нет необходимости отдельно обрабатывать случаи головы списка и внутренних узлов.

**4. Является ли одна строка исходной для другой**

В исходную строку добавили некоторое количество символов.
Необходимо определить, была ли строка a исходной для строки b.

Реализуйте алгоритм сравнения строк с использованием указателей. Объясните, как
вы определяете, что вторая строка получена добавлением символов к первой.


```python
def isSubsequence(a, b):
    i, j = 0, 0
    while (i < len(a)) and (j < len(b)):
        if a[i] == b[j]:
            i += 1
            j += 1
        else:
            j += 1
    return i == len(a)
print(isSubsequence('abc', 'faxcbhjcc'))
```

Создаем по указателю для каждой строки. Если символы на указателях равны, сдвигаем оба указателя. Иначе сдвигаем только указатель строки b. В конце проверяем, дошел ли указатель строки а до конца.

**5. Метод двух указателей (поиск пары)**

Дан отсортированный массив чисел и целевое число target.
Необходимо определить, существуют ли два элемента, сумма которых равна target.

Реализуйте решение с использованием двух указателей (один в начале, другой в
конце массива). Поясните, почему такой подход работает за O(n)

```python
def TwoSum(arr, target): # Возвращаем -1, если таких чисел нет
    if ((arr[0] + arr[1]) > target) or ((arr[-1] + arr[-2]) < target): # Не имеет смысл проводить алгоритм, т.к сумма двух наименьших больше таргета или сумма двух наибольших меньше таргета
        return -1
    left = 0
    right = len(arr) - 1
    while left < right:
        sum = arr[left] + arr[right]
        if sum == target:
            return [left, right]
        elif sum > target:
            right -= 1
        else:
            left += 1
    return -1
print(TwoSum([3, 8, 9, 11, 16, 18, 19, 21], 25))
print(TwoSum([3, 8, 9, 11, 16, 18, 19, 21], 100))
```

В цикле не более n итераций (в каждой итерации либо нашли пару, либо сдвинули указатель). Отсюда сложность O(N).

**6. Является ли слово палиндромом?**

Напишите функцию, которая принимает строку и возвращает true, если она является
палиндромом.
В противном случае вернуть false.

Реализуйте проверку с помощью двух указателей. Объясните, как сравниваются
символы и почему алгоритм работает за линейное время.

```python
def isPalindrome(line):
    left = 0
    right = len(line) - 1
    while left < right:
        if line[left] != line[right]:
            return False
        else:
            left += 1
            right -= 1
    return True
print(isPalindrome('asdfdsa'))
print(isPalindrome('asdfgdsa'))
```

Сложность по аналогии с прошлой задачей. Символы сравниваются зеркально середине строки (два указателя с начала и с конца)

**7. Метод двух указателей (удаление дубликатов)**

Дан отсортированный массив.
Необходимо удалить дубликаты in-place так, чтобы каждый элемент встречался только
один раз.

Реализуйте алгоритм с двумя указателями (медленный и быстрый). Поясните, как
обеспечивается перезапись дубликатов без использования дополнительной памяти

```python
def deleteDuplicates(arr):
    slow = 0
    for fast in range (1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]
    slow += 1
    return arr[:slow]
print(deleteDuplicates([1, 1, 2, 5, 7, 7, 7, 9, 11, 11]))
print(deleteDuplicates([1]))
print(deleteDuplicates([1, 1, 1, 1, 1, 1]))
```

slow - указывает на последний элемент уникальной части массива

fast - пробегает по всему массиву

Как только находим уникальный элемент сдвигаем slow и записываем на его место новое уникальное значение

**8. Слияние двух отсортированных списков**

Написать функцию, которая принимает два отсортированных односвязных списка и
объединяет их в один отсортированный список.
Затраты по памяти должны быть O(1).

Реализуйте слияние, изменяя только ссылки между существующими узлами.
Объясните, почему алгоритм работает за O(n + m) и не требует дополнительной
памяти.

```python
def mergeLinkedLists(head1, head2):
    dummy = Node(0)
    tail = dummy

    p1, p2 = head1, head2

    while p1 is not None and p2 is not None:
        if p1.value <= p2.value:
            tail.next = p1
            p1 = p1.next
        else:
            tail.next = p2
            p2 = p2.next
        tail = tail.next

    tail.next = p1 if p1 is not None else p2

    return dummy.next
list1 = list_to_linked_list([1, 5, 7, 11])
list2 = list_to_linked_list([3, 4, 8, 9, 12])
print_linked_list(mergeLinkedLists(list1, list2))
```
