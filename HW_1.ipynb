**1. Two sum**

Дан отсортированный по возрастанию массив целых чисел
и некоторое число target.

Необходимо найти два числа в массиве, которые в сумме дают
заданное значение target, и вернуть их индексы.



```python
def TwoSum(arr, target): # Возвращаем -1, если таких чисел нет
    if ((arr[0] + arr[1]) > target) or ((arr[-1] + arr[-2]) < target): # Не имеет смысл проводить алгоритм, т.к сумма двух наименьших больше таргета или сумма двух наибольших меньше таргета
        return -1
    left = 0
    right = len(arr) - 1
    while left < right:
        sum = arr[left] + arr[right]
        if sum == target:
            return [left, right]
        elif sum > target:
            right -= 1
        else:
            left += 1
    return -1
print(TwoSum([3, 8, 9, 11, 16, 18, 19, 21], 25))
print(TwoSum([3, 8, 9, 11, 16, 18, 19, 21], 100))
```

**2. Развернуть массив**

Дан массив целых чисел.

Необходимо развернуть его.

Сделать это надо за линейное время без дополнительных
аллокаций.

```python
def reverseArray(arr):
    left = 0
    right = len(arr) - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr
print(reverseArray([1, 2, 3, 4, 5]))
```

**3. Развернуть часть массива**

Дан массив целых чисел.

Необходимо повернуть (сдвинуть) справа налево
часть массива, которая указана вторым параметром.

Сделать это надо за линейное время
без дополнительных аллокаций


```python
def reverseArray(arr, left, right):
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

def reverseArrayPart(arr, k):
    n = len(arr)
    k = k % n
    if k == 0:
        return arr
    arr = reverseArray(arr, 0, n-1)
    arr = reverseArray(arr, 0, k-1)
    arr = reverseArray(arr, k, n-1)
    return arr

print(reverseArrayPart([1, 2, 3, 4, 5, 6, 7], 3))
```

**4. Слияние двух отсортированных массивов**

Дано два отсортированных массива.

Необходимо написать функцию которая объединит
эти два массива в один отсортированный.

```python
def mergeArrays_1(arr1, arr2):
    i = 0
    j = 0
    arr = []
    while (i < len(arr1)) and (j < len(arr2)):
        if arr1[i] < arr2[j]:
            arr.append(arr1[i])
            i += 1
        else:
            arr.append(arr2[j])
            j += 1
    while i < len(arr1):
        arr.append(arr1[i])
        i += 1
    while j < len(arr2):
        arr.append(arr2[j])
        j += 1
    return arr
print(mergeArrays_1([3, 8, 10, 11], [1, 7, 9]))
```

**5. Слияние двух отсортированных массивов без дополнительных аллокаций**

Дано два отсортированных массива.

Необходимо написать функцию которая
объединит эти два массива в один
отсортированный.

Первый массив имеет размер, равный
результирующиму массиву, значения
в конце равны нулям. Мы не должны
создавать третий массив.


```python
def mergeArrays_2(arr1, arr2):
    k = len(arr1) - 1
    j = len(arr2) - 1
    i = k - j - 1
    while k >= 0:
        if (i >= 0) and (arr1[i] > arr2[j]):
                arr1[k] = arr1[i]
                i -= 1
        elif j >= 0:
            arr1[k] = arr2[j]
            j -= 1
        k -= 1
    return arr1
print(mergeArrays_2([3, 8, 10, 11, 0, 0, 0], [1, 7, 9]))
print(mergeArrays_2([1, 8, 10, 11, 0, 0, 0], [3, 7, 9]))
```

**6. Минимальный размер подмассива**

Дан массив положительных целых чисел nums и
положительное целое число target. Минимальную длину подмассива, сумма элементов
которого больше или равна target. Если такого
подмассива не существует, верните 0.

```python
def minSubArray(arr, target):
    left = 0
    right = 0
    minLen = len(arr) + 1
    sum = arr[left]
    while True:
        if sum == target: # Числа положительные, поэтому можем сдвинуть окно целиком на 1 вправо
            minLen = min(right - left + 1, minLen)
            right += 1
            if right == len(arr):
                break
            sum -= arr[left]
            left += 1
            sum += arr[right]
        elif sum < target:
            right += 1
            if right == len(arr):
                break
            sum += arr[right]
        else:
            sum -= arr[left]
            left += 1
    if minLen == (len(arr) + 1):
        return 0
    return minLen

print(minSubArray([2,3,1,2,4,3], 7))
print(minSubArray([1,4,4], 4))
print(minSubArray([1,1,1,1,1,1,1,1], 11))
print(minSubArray([2], 2))
print(minSubArray([2], 1))
```

**7. Сортировка массива из 0 и 1**

Дан массив, содержащий только
0 и 1.

Напишите функцию, которая
сортирует его так, чтобы все нули
оказались в начале, а все единички —
в конце.


```python
def sort_bin_array(arr):
    left = 0
    right = len(arr) - 1
    while (left < right):
        if arr[left] == 1:
            arr[left], arr[right] = arr[right], arr[left]
            right -= 1
        elif arr[right] == 0:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
        else:
            right -= 1
            left += 1
    return arr
print(sort_bin_array([0, 1, 1, 0, 0, 1, 1, 1, 0, 1]))
```

**8. Задача флага Нидерландов**

Дан массив состоящий из нулей,
единиц и двоек.

Необходимо его отсортировать
за линейное время


```python
def sortColors(arr):
    left = 0
    mid = 0
    right = len(arr) - 1
    while mid <= right:
        if arr[mid] == 2:
            arr[mid], arr[right] = arr[right], arr[mid]
            right -= 1
        elif arr[mid] == 0:
            arr[left], arr[mid] = arr[mid], arr[left]
            left += 1
            mid += 1
        else:
            mid += 1
    return arr
print(sortColors([2, 0, 1, 1, 0, 2, 1, 1, 0]))
```

**9. Передвинуть четные числа вперед**

Дан не отсортированный массив целых чисел.

Необходимо перенести в начало массива все
четные числа, сохраняя их очередность


```python
def evenFirst(arr):
    left = 0
    for right in range(0, len(arr)):
        if arr[right] % 2 == 0:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
    return arr
print(evenFirst([2, 5, 7, 9, 8 , 4, 13, 11, 12]))
```

**10. Нули в конец**

Есть массив с оценками, среди
которых есть 0. Необходимо все оценки,
равные нулю перенести в конец массива,
чтобы все такие школьники оказались
в конце списка.

```python
def zero_at_the_end(arr):
    is_zero = 0
    while arr[is_zero] != 0:
        is_zero += 1
    for non_zero in range(is_zero + 1, len(arr)):
        while arr[is_zero] != 0:
            is_zero += 1
        if arr[non_zero] == 0:
            continue
        else:
            arr[non_zero], arr[is_zero] = arr[is_zero], arr[non_zero]
    return arr
print(zero_at_the_end([0, 0, 1, 0, 3, 12]))
print(zero_at_the_end([2, 0, 1, 0, 3, 12]))
print(zero_at_the_end([0 , 0, 0, 18, 16, 0, 0, 77, 99]))
```
